```python
import pandas as pd
import numpy as np
```

```python
master_df=pd.read_csv(r"D:\RCM\Cleaned data\Master_df.xls")
```

# Master KPI Table

```python
KPI_Table = pd.DataFrame(columns=["Category", "KPI","Value"])
```

```python
KPI_Table
```

# KPI Function

```python
def add_kpi(category, kpi, value):

    global KPI_Table

    KPI_Table.loc[len(KPI_Table)] = [category,kpi,value]
```

# ________________________ Executive KPIs _________________________

```python
add_kpi("Executive","Total Claims",master_df["Claim_ID"].nunique())

add_kpi("Executive","Total Revenue",round(master_df["Net_Realized_Revenue"].sum(),2))

add_kpi("Executive","Total Collections",round(master_df["Total_Collections_Posted"].sum(),2))

add_kpi("Executive","Outstanding Amount",round(master_df["Outstanding_Amount"].sum(),2))

add_kpi("Executive","Net Profit",round(master_df["Net_Profit"].sum(),2))

add_kpi("Executive","Profit Margin %",round(master_df["Profit_Margin"].mean(),2))

add_kpi("Executive","Collection %",round((master_df["Total_Collections_Posted"].sum() /master_df["Net_Billed_Amount"].sum()) * 100,2))

add_kpi("Executive","Approval %",round((master_df["Approval_Flag"]=="Approved").mean()*100, 2))

add_kpi("Executive","Denial %",round((master_df["Denial_Flag"]=="Yes").mean()*100,2 ))

add_kpi("Executive","Average Settlement Days",round(master_df["Days_to_Payment_Settlement"].mean(),2))
```

```python
KPI_Table
```

# ___________________________ Revenue KPIs ___________________________

```python
add_kpi("Revenue","Total Net Billed Amount",round(master_df["Net_Billed_Amount"].sum(),2))

add_kpi("Revenue","Total Net Billed Amount",round(master_df["Net_Billed_Amount"].sum(),2))

add_kpi("Revenue","Average Revenue per Claim",round(master_df["Net_Realized_Revenue"].sum()/master_df["Claim_ID"].nunique(),2))

add_kpi("Revenue","Average Revenue per Patient",round(master_df["Net_Realized_Revenue"].sum()/master_df["Patient_ID"].nunique(),2))

add_kpi("Revenue","Highest Revenue Department",master_df.groupby("Department_Name")["Net_Realized_Revenue"].sum().idxmax())

add_kpi("Revenue","Highest Revenue Insurance Company",master_df.groupby("Insurance_Company")["Net_Realized_Revenue"].sum().idxmax())
```

```python
KPI_Table[KPI_Table["Category"]=="Revenue"]
```

# _____________________________ Collection KPIs ________________________________

```python
add_kpi("Collection","Outstanding Amount",round(master_df["Outstanding_Amount"].sum(),2))


add_kpi("Collection","Collection Gap",round((
            master_df["Net_Billed_Amount"] -master_df["Total_Collections_Posted"]).sum(), 2 ))


add_kpi("Collection","Average Collection per Claim",round(
        master_df["Total_Collections_Posted"].sum()/master_df["Claim_ID"].nunique(),2))


add_kpi("Collection","Average Collection per Patient",round(
        master_df["Total_Collections_Posted"].sum()/master_df["Patient_ID"].nunique(),2))


add_kpi("Collection","Average Payment Settlement Days",round(master_df["Days_to_Payment_Settlement"].mean(),2))


add_kpi("Collection","Pending Collection %",round((
            master_df["Outstanding_Amount"].sum()/master_df["Net_Billed_Amount"].sum())*100,2))
```

```python
KPI_Table[KPI_Table["Category"]=="Collection"]
```

# ________________________________ Claim KPIs ______________________________

```python
add_kpi("Claim","Total Claims Submitted",master_df["Claim_ID"].nunique())

add_kpi("Claim","High Value Claim %",round((master_df["High_Value_Claim"]=="Yes").mean()*100,2))

add_kpi("Claim","Average Submission Attempts",round(master_df["Submission_Attempts_Count"].mean(),2))

add_kpi("Claim","Average Claim Processing Time (Hours)",round(master_df["Case_Processing_Time_Hours"].mean(),2))
```

```python
KPI_Table[KPI_Table["Category"]=="Claim"]
```
