# ____________________ IQR Method (Count Outliers) _______________________

## Function to Count Outliers

```python
def count_outliers(column):

    Q1 = master_df[column].quantile(0.25)
    Q3 = master_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = master_df[(master_df[column] < lower) |
                        (master_df[column] > upper)]

    print(f"{column}")
    print("Total Outliers :", len(outliers))
    print("-"*40)
```

```python
count_outliers("Net_Realized_Revenue")

count_outliers("Total_Collections_Posted")

count_outliers("Outstanding_Amount")

count_outliers("Net_Profit")

count_outliers("Profit_Margin")

count_outliers("Days_to_Payment_Settlement")
```
