# Create Cleaning Reusable Functions for automation

# Function 1 : Missing Values

```python
def clean_missing_values(df):

    return df.dropna(how="all")
```

# Function 2 : Duplicate Records

```python
def remove_duplicates(df):

    return df.drop_duplicates()
```

# Function 3 : Standardize Missing Values 
### (NULL ,null , N/A , NA , - , Blank , Unknown)

```python
def standardize_missing_values(df):

    return df.replace(
        ["NULL", "null", "N/A", "NA", "-", ""],
        pd.NA)
```

# Function 4 :Convert Date Columns

```python
def convert_dates(df):

    for col in df.columns:

        if "Date" in col:

            df[col] = pd.to_datetime(
                df[col],
                errors="coerce")

    return df
```

# Function 5 — Clean Text Columns

```python
def clean_text_columns(df):

    text_columns = df.select_dtypes(include="object").columns

    for col in text_columns:
        df[col] = df[col].str.strip()

    return df
```

# Function 6 — Standardize Column Names

```python
def clean_column_names(df):

    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
    )

    return df
```

# Merge all cleaning functions in one 

```python
def clean_table(df):

    df = standardize_missing_values(df)

    df = clean_missing_values(df)

    df = remove_duplicates(df)

    df = clean_text_columns(df)

    df = standardize_missing_values(df)

    df = convert_dates(df)

    return df
```

# Test on One Table

# Cleaning Loop (21 Tables)

```python
clean_tables = {}

for table_name, df in tables.items():

    clean_tables[table_name] = clean_table(df)
```

```python
len(clean_tables)
```

# Clean All Tables

```python
clean_tables = {}

cleaning_summary = []

for table_name, df in tables.items():

    rows_before = df.shape[0]

    df_clean = clean_table(df)

    rows_after = df_clean.shape[0]

    clean_tables[table_name] = df_clean

    cleaning_summary.append({

        "Table Name": table_name,

        "Rows Before": rows_before,

        "Rows After": rows_after,

        "Rows Removed": rows_before - rows_after,

        "Status": "Cleaned"

    })
```

# Cleaning Summary Report

```python
cleaning_summary = pd.DataFrame(cleaning_summary)

cleaning_summary
```

```python
for table_name, df in clean_tables.items():

    df.to_csv(
        f"D:/RCM/Cleaned Data/{table_name}.csv",
        index=False
    )

print("All Cleaned Tables Saved Successfully.")
```
